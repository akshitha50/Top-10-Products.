```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("sample_-_superstore.csv.xls")

df.head()
```

```python


top_10 = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)


print(top_10)
```

```python

top_10.sort_values().plot(
    kind="barh",
    figsize=(10, 6),
    title="Top 10 Products by Sales"
)

plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
top_10_df = top_10.reset_index()
top_10_df.columns = ["Product Name", "Total Sales"]

top_10_df.to_csv("Top_10_Products_Table.csv", index=False)

print("Top 10 products table saved successfully!")
```
